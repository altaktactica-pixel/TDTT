Bài 2: Một khu đất hình chữ nhật có chiều rộng a, chiều dài b. Người ra xây một khu
vui chơi hình tròn tại vị trí trung tâm khu đất như hình vẽ. Nhập vào a, b. Hãy tính phần
diện tích còn để trồng cây xung quanh là bao nhiêu. Biết pi = 3.14, lấy kết quả đến 2
chữ số phần thập phân. 


```python
a = float(input())
b = float(input())

S_hcn = a * b
r = a / 2
S_tron = 3.14 * (r ** 2)

S_con_lai = S_hcn - S_tron

print(f"Diện tích còn lại: {S_con_lai:.2f}")
```

Bài 6: Nhập vào 3 số a, b, c. Kiểm tra xem đây có phải là cạnh của 1 tam giác không,
nếu đúng thì tính diện tích của tam giác đó, in ra với 1 chữ số phần thập phân, nếu nó
không phải 3 cạnh của tam giác thì in ra “Khong phai 3 canh cua tam giac”. 

```python
a = float(input())
b = float(input())
c = float(input())

if a + b > c and a + c > b and b + c > a:
    p = (a + b + c) / 2
    
    dien_tich = (p * (p - a) * (p - b) * (p - c)) ** 0.5
    
    print(f"Dien tich tam giac la: {dien_tich:.1f}")
else:
    print("Khong phai 3 canh cua tam giac")
```

Bài 8: : Nhập vào họ tên chủ hộ và chỉ số điện kế tháng trước và chỉ số điện kế tháng này,
tính tiền điện tháng này cho hộ biết:
- Từ 0 – 50kWh: 1.984 đồng/kWh
- Bậc 2: Từ 51 – 100 kWh: 2.050 đồng/kWh
- Bậc 3: Từ 101 – 200 kWh: 2.380 đồng/kWh
- Bậc 4: Từ 201 – 300 kWh: 2.998 đồng/kWh
- Bậc 5: Từ 301 – 400 kWh: 3.350 đồng/kWh
- Bậc 6: Từ 401 trở đi kWh: 3.460 đồng/kWh 


```python

ten_chu_ho = input("Ten chu ho: ")
chi_so_truoc = int(input("Chi so thang truoc: "))
chi_so_nay = int(input("Chi so thang nay: "))


kwh = chi_so_nay - chi_so_truoc
tien_dien = 0


if kwh <= 50:
    tien_dien = kwh * 1984
elif kwh <= 100:
    tien_dien = 50 * 1984 + (kwh - 50) * 2050
elif kwh <= 200:
    tien_dien = 50 * 1984 + 50 * 2050 + (kwh - 100) * 2380
elif kwh <= 300:
    tien_dien = 50 * 1984 + 50 * 2050 + 100 * 2380 + (kwh - 200) * 2998
elif kwh <= 400:
    tien_dien = 50 * 1984 + 50 * 2050 + 100 * 2380 + 100 * 2998 + (kwh - 300) * 3350
else:
    tien_dien = 50 * 1984 + 50 * 2050 + 100 * 2380 + 100 * 2998 + 100 * 3350 + (kwh - 400) * 3460


tien_phai_tra = round(tien_dien * 1.08)


print(f"Ho va ten: {ten_chu_ho}")
print(f"Tien phai tra la: {tien_phai_tra}")
```

Bài 7: Bài tập 5: Tính điểm học lực
Nhập điểm (0-10), phân loại học lực theo quy tắc:
>=8: Giỏi; >=6.5 và <8: Khá; >=5 và <6.5: Trung bình; <5: Yếu 


```python
n = float(input())
if n < 0 or n > 10:
    print("Điểm không hợp lệ")
else:
    if n >= 8:
        print("Học sinh giỏi")
    elif n >= 6.5:
        print("Học sinh khá")
    elif n >= 5:
        print("Học sinh trung bình")
    else:
        print("Học sinh yếu")
```

Bài 8: Kiểm tra năm nhuận: Viết chương trình nhập năm, kiểm tra xem năm đó có phải

nhưng không chia hết cho 100)

```python
nam = int(input())

if (nam % 400 == 0) or (nam % 4 == 0 and nam % 100 != 0):
    print(f"Năm {nam} là năm nhuận.")
else:
    print(f"Năm {nam} không phải là năm nhuận.")
```
