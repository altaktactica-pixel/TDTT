W5A16:Viết hàm kiểm tra một số có phải số Armstrong hay không? Hàm trả về kiểu bool, có tham số là một số 
nguyên x 

```python
def is_armstrong(x: int) -> bool:
    if x < 0:
        return False
    chuoi_x = str(x)
    so_mu = len(chuoi_x)
    tong = 0
    for chu_so in chuoi_x:
        tong += int(chu_so) ** so_mu
    return tong == x
n=int(input())
if is_armstrong(n):
    print("True")
else:
    print("False")
```

```python
def cp(x:int)->bool:
    if x<0:
        return False
    for i in range(1,int(x**0.5)+1):
        if i*i==x:
            return True
    return False
n=int(input())
if cp(n):
    print("True")
else:
    print("False")
```

W5A19: Viết hàm kiểm tra một số có phải số kỳ diệu hay không, xử lý trên dữ liệu kiểu số nguyên.
(số kì diệu là nếu ta cộng dồn các chữ số của nó lại nhiều lần, cho đến khi chỉ còn 1 chữ số, và kết quả rút gọn cuối cùng phải bằng 1.)

```python
def is_magic_number(x: int) -> bool:
    x = abs(x)
    while x > 9:
        tong = 0
        while x > 0:
            tong += x % 10
            x = x // 10
        x = tong
    return x == 1
n = int(input())
if is_magic_number(n):
    print("True")
else:    print("False")
    
```

W5A20: Viết hàm detect_number_type(s) trả về 'int' nếu s là số nguyên, 'float' nếu là số thực, ngược lại 'neither'. 

```python
def detect_number_type(s):
    # Lấy chiều dài của chuỗi
    do_dai = len(s)
    
    # Bẫy 1: Người dùng không nhập gì cả
    if do_dai == 0:
        return 'neither'

    dem_dau_cham = 0
    vi_tri_bat_dau = 0

    # Nếu ký tự đầu tiên (vị trí 0) là dấu trừ, ta sẽ xét từ vị trí 1
    if s[0] == '-':
        vi_tri_bat_dau = 1
        
        # Bẫy 2: Nhỡ người dùng chỉ gõ mỗi một dấu "-"
        if do_dai == 1:
            return 'neither'

    # Dùng vòng lặp soi từng ký tự một
    for i in range(vi_tri_bat_dau, do_dai):
        ky_tu = s[i]

        # Nếu là dấu chấm thì đếm
        if ky_tu == '.':
            dem_dau_cham += 1
            
        # Nếu ký tự không nằm trong danh sách các con số 
        elif ky_tu not in "0123456789":
            return 'neither'  # Phát hiện chữ cái linh tinh -> Báo rác luôn!

    # --- CHỐT KẾT QUẢ CUỐI CÙNG ---
    if dem_dau_cham == 0:
        return 'int'
        
    elif dem_dau_cham == 1:
        # Bẫy 3: Nhỡ người dùng nhập mỗi dấu "." hoặc "-."
        if do_dai == 1 or (do_dai == 2 and s[0] == '-'):
            return 'neither'
        return 'float'
        
    else:
        return 'neither'  # Trường hợp 2, 3 dấu chấm...

# --- KIỂM THỬ ---
print(detect_number_type("-45"))      # int
print(detect_number_type("3.14"))     # float
print(detect_number_type("12.34.56")) # neither
print(detect_number_type("abc"))      # neither
```

W5A23: Viết hàm quadratic_roots_count(a, b, c) trả về số nghiệm thực của phương trình a*x^2 + b*x + c = 0. 

```python
def quadratic_roots_count(a, b, c):
    delta = b**2 - 4*a*c
    if delta > 0:
        return 2
    elif delta == 0:
        return 1
    else:
        return 0

a, b, c = map(float, input().split())
so_nghiem = quadratic_roots_count(a, b, c)

if so_nghiem == 2:
    print("Phương trình có 2 nghiệm")
elif so_nghiem == 1:
    print("Phương trình có 1 nghiệm")
else:
    print("Phương trình vô nghiệm")
```

W5A24: Viết hàm are_collinear(x1,y1,x2,y2,x3,y3) kiểm tra ba điểm có thẳng hàng hay không. 

```python
def are_collinear(x1, y1, x2, y2, x3, y3):
    # Nếu diện tích tam giác = 0 thì 3 điểm thẳng hàng
    dien_tich = x1*(y2 - y3) + x2*(y3 - y1) + x3*(y1 - y2)
    return dien_tich == 0

x1, y1, x2, y2, x3, y3 = map(float, input().split())

if are_collinear(x1, y1, x2, y2, x3, y3):
    print("True")
else:
    print("False")
```

W5A25: Viết hàm point_in_quadrilateral(quad, p) kiểm tra điểm p có nằm trong hoặc trên biên của tứ giác quad hay 
không. Trong đó: quad là một tứ giác (4 đỉnh – 4 điểm), p là một điểm cần kiểm tra. VD: quad = [(0,0), (4,0), (4,3), 
(0,3)], p = (2,1) 

```python
def point_in_quadrilateral(quad, p):
    x, y = p
    inside = False
    p1x, p1y = quad[0]
    # Thuật toán kiểm tra điểm trong đa giác
    for i in range(1, 5):
        p2x, p2y = quad[i % 4]
        if y > min(p1y, p2y):
            if y <= max(p1y, p2y):
                if x <= max(p1x, p2x):
                    if p1y != p2y:
                        xints = (y - p1y) * (p2x - p1x) / (p2y - p1y) + p1x
                    if p1x == p2x or x <= xints:
                        inside = not inside
        p1x, p1y = p2x, p2y
    return inside

quad = [(0,0), (4,0), (4,3), (0,3)]
px, py = map(float, input().split())
p = (px, py)

if point_in_quadrilateral(quad, p):
    print("Điểm nằm trong tứ giác")
else:
    print("Điểm nằm ngoài tứ giác")
```

W5A27:  Viết hàm trả về số lượng chữ số khác 0 trong n số nguyên nhập vào từ bàn phím (một list) 

```python
def dem_chu_so_khac_0(danh_sach):
    dem = 0
    # Xét từng số trong mảng
    for so in danh_sach:
        # Ép qua chuỗi để soi từng chữ số cho dễ (giống các bài trước)
        chuoi_so = str(abs(so))
        for chu_so in chuoi_so:
            if chu_so != '0':
                dem += 1
    return dem

# Cách nhập mảng siêu tốc trên 1 dòng
danh_sach = [int(x) for x in input().split()]
print(dem_chu_so_khac_0(danh_sach))
```

W5A28: Viết hàm tính góc ∠ABC (độ) từ tọa độ A(x1,y1), B(x2,y2), C(x3,y3). 

```python
import math

def tinh_goc_ABC(x1, y1, x2, y2, x3, y3):
    # Tính vector BA và BC
    ba_x, ba_y = x1 - x2, y1 - y2
    bc_x, bc_y = x3 - x2, y3 - y2
    
    # Tính tích vô hướng và độ dài 2 cạnh
    tich_vo_huong = ba_x * bc_x + ba_y * bc_y
    do_dai_BA = math.sqrt(ba_x**2 + ba_y**2)
    do_dai_BC = math.sqrt(bc_x**2 + bc_y**2)
    
    # Công thức: cos(B) = (BA * BC) / (|BA| * |BC|)
    cos_B = tich_vo_huong / (do_dai_BA * do_dai_BC)
    
    # Đổi từ Cos ra góc (Radian) rồi đổi tiếp sang Độ (Degrees)
    goc_radian = math.acos(cos_B)
    return math.degrees(goc_radian)

x1, y1, x2, y2, x3, y3 = map(float, input().split())
print(tinh_goc_ABC(x1, y1, x2, y2, x3, y3))
```

W5A30: Viết hàm tính hệ số nhị thức C(n, k).

```python
import math

def to_hop_chap_k_cua_n(n, k):
    # Công thức: C(n, k) = n! / (k! * (n-k)!)
    tu_so = math.factorial(n)
    mau_so = math.factorial(k) * math.factorial(n - k)
    
    # Dùng chia nguyên (//) vì hệ số luôn là số nguyên
    return tu_so // mau_so

n, k = map(int, input().split())
print(to_hop_chap_k_cua_n(n, k))
```

W5A33: Viết hàm smart_taxi_fare(car_type, start_hour, km, wait_minutes) tính tổng tiền taxi theo quy tắc giá mở cửa, 
giá km, phí chờ, phụ phí ban đêm và giảm giá. 

```python
def smart_taxi_fare(car_type, start_hour, km, wait_minutes):
    # 1. GIÁ MỞ CỬA & GIÁ KM (Giả sử: xe 4 chỗ và xe 7 chỗ)
    if car_type == 4:
        gia_mo_cua = 10000
        gia_km = 12000
    else: # Xe 7 chỗ
        gia_mo_cua = 12000
        gia_km = 15000
        
    # 2. PHÍ CHỜ (Giả sử: 2000đ / 1 phút)
    phi_cho = wait_minutes * 2000
    
    # Cộng dồn tiền cơ bản
    tong_tien = gia_mo_cua + (km * gia_km) + phi_cho
    
    # 3. PHỤ PHÍ BAN ĐÊM (Giả sử: Đi từ 22h đêm đến 5h sáng bị thu thêm 15% phí)
    if start_hour >= 22 or start_hour <= 5:
        tong_tien = tong_tien * 1.15  # Nhân 1.15 tức là cộng thêm 15%
        
    # 4. GIẢM GIÁ (Giả sử: Đi xa trên 20km thì được giảm 10% tổng hóa đơn)
    if km > 20:
        tong_tien = tong_tien * 0.9   # Nhân 0.9 tức là chỉ thu 90% tiền
        
    return tong_tien

car_type = int(input("Nhập loại xe (4 hoặc 7): "))
start_hour = int(input("Nhập giờ đi (0-24): "))
km = float(input("Nhập số km: "))
wait_minutes = int(input("Nhập thời gian chờ (phút): "))

tien_taxi = smart_taxi_fare(car_type, start_hour, km, wait_minutes)
print(f"Tổng tiền: {tien_taxi}")
```

W5A34:Viết hàm sum_1_to_n(n) trả về tổng các số x từ 1 đến n, nếu x có các chữ số đôi một khác nhau 
Nếu n ≤ 0 trả về 0.

```python
def sum_1_to_n(n):
    # Đề bài yêu cầu: Nếu n <= 0 thì trả về 0
    if n <= 0:
        return 0
        
    tong = 0
    for x in range(1, n + 1):
        # Ép số x thành chuỗi để soi từng chữ số
        chuoi_x = str(x)
        
        # Kiểm tra: Độ dài của chuỗi gốc có bằng độ dài của tập hợp set() không?
        # Nếu bằng nhau -> Không có số nào bị trùng -> Cộng vào tổng
        if len(chuoi_x) == len(set(chuoi_x)):
            tong += x
            
    return tong
n = int(input())
print(sum_1_to_n(n))
```
