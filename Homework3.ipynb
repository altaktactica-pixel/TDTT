Bài 4:  Viết chương trình in ra hình tam giác cân có độ cao h được nhập từ bàn phím 
Ví dụ: Ta nhập h = 4: 


```python
h = int(input())

for i in range(1, h + 1):
    if i == 1:
        # Dòng đỉnh tam giác
        print(" " * (h - i) + "*")
    elif i == h:
        # Dòng đáy tam giác
        print("*" * (2 * h - 1))
    else:
        # Các dòng rỗng ở giữa
        khoang_trang_le = " " * (h - i)
        khoang_trang_giua = " " * (2 * i - 3)
        print(khoang_trang_le + "*" + khoang_trang_giua + "*")
```

Bài 9:  Nhập vào số nguyên dương a, kiểm tra xem a có phải là số chính phương hay 
không? Số chính phương là số có căn bậc hai là một số nguyên VD: 1, 4, 9, 16, 25….

```python
while True:
    a = int(input())
    if a > 0:
        break
for i in range(1, int(a**0.5)+1):
    if a == i**2:
        print(f"{a} là số chính phương")
        break
else:
    print(f"{a} không phải là số chính phương")
```

Bài 14: Nhập vào một dãy số nguyên, dừng nhập khi người dùng nhập -1. Sau khi nhập 
xong, in số lớn nhất, số nhỏ nhất trong những số vừa nhập. KHÔNG ĐƯỢC dùng mảng

```python
# 1. Lấy dữ liệu người dùng nhập vào trên 1 dòng
# Lệnh split() sẽ băm câu "5 12 8 -1" thành các chữ rời rạc: ['5', '12', '8', '-1']
day_so = input().split()

co_cot_moc = False
lon_nhat = 0
nho_nhat = 0

# 2. Vòng lặp bốc từng chữ trong dãy vừa băm ra để xét
for chu_so in day_so:
    
    # Ép chữ (ví dụ '5') thành số nguyên (5) để làm toán được
    n = int(chu_so) 
    
    # Nếu gặp đúng số -1 thì đập vỡ vòng lặp, nghỉ chơi!
    if n == -1:
        break
        
    # 3. Kỹ thuật nhớ kỷ lục (không cần mảng)
    if co_cot_moc == False:
        # Người đầu tiên bước ra mặc định ôm cả 2 kỷ lục
        lon_nhat = n
        nho_nhat = n
        co_cot_moc = True  # Đánh dấu là đã có kỷ lục để tí khỏi nhầm
    else:
        # Từ người thứ 2 trở đi, đem đi so kè
        if n > lon_nhat:
            lon_nhat = n  # Ghi nhận kỷ lục lớn mới
            
        if n < nho_nhat:
            nho_nhat = n  # Ghi nhận kỷ lục nhỏ mới

# 4. Vòng lặp kết thúc, in sổ ghi chép ra
if co_cot_moc == True:  # Chắc chắn là có người nhập, chứ không phải vừa vào đã gõ -1
    print(lon_nhat)
    print(nho_nhat)
```

Bài 16: Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số chẵn, bao nhiêu 
chữ số lẻ

```python
n = int(input())
dem_chan = 0
dem_le = 0

while n > 0:
    chu_so = n % 10  
    
    if chu_so % 2 == 0:
        dem_chan += 1
    else:
        dem_le += 1
        
    n = n // 10      

print("Chẵn:", dem_chan)
print("Lẻ:", dem_le)
```

Bài 18:  Nhập vào một số nguyên dương n, kiểm tra xem n có phải là số dạng 3^k hay 
không

```python
n = int(input())

if n <= 0:
    print("Không phải")
else:
    while n % 3 == 0:
        n = n // 3
    if n == 1:
        print("Đúng, là số dạng 3^k")
    else:
        print("Không phải")
```

Bài 19: Dãy số fibonacci là dãy số được định nghĩa như sau: 1, 1, 2, 3, 5, 8, 13,... với số 
kế tiếp sẽ bằng tổng hai số trước đó. Nhập vào số nguyên dương A, hãy tìm số trong 
dãy số fibonacci lớn nhất nhưng không vượt quá A 

```python
A = int(input())

f1 = 1  
f2 = 1  
ky_luc = 0 
while f1 <= A:
    ky_luc = f1
    f3 = f1 + f2
    f1 = f2
    f2 = f3
print(ky_luc)
```

Bài 20:Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số là số nguyên tố.

```python
n = int(input())
dem_nguyen_to = 0

while n > 0:
    chu_so = n % 10  
    if chu_so == 2 or chu_so == 3 or chu_so == 5 or chu_so == 7:
        dem_nguyen_to += 1  
    n = n // 10      

print(dem_nguyen_to)
```
