*1.Yêu cầu. Cước phí Taxi Thông Minh. 
Viết một hàm duy nhất smart_taxi_fare(car_type, start_hour, km, wait_minutes) tính và trả về total (tổng tiền phải trả, 
kiểu số nguyên, đơn vị VND). 
 
Các quy tắc cần cài đặt: 
1. Giá mở cửa (0–0.8 km): - Xe 4 chỗ: 11,000 VND - Xe 7 chỗ: 12,000 VND 
2. Giá theo km sau 0.8 km: - Xe 4 chỗ: từ 0.8→20 km nh 13,000 VND/km; trên 20 km tính 10,500 VND/km - Xe 7 chỗ: từ 0.8→20 km nh 14,000 VND/km; trên 20 km tính 12,000 VND/km 
3. Phí chờ: 5 phút đầu miễn phí; sau đó, mỗi 5 phút (làm tròn lên theo từng block 5 phút) tính 2,000 VND. 
4. Phụ thu ban đêm: nếu start_hour thuộc [22,23,0,1,2,3,4,5], cộng +20% trên tiền tạm tính (mở cửa + quãng đường + 
chờ). 
5. Điều kiện đặc biệt: - Nếu km > 30 và không phải ban đêm, áp dụng giảm 10% trên tổng tiền. - Nếu tổng tiền cuối cùng < 50,000 nhưng km > 5, đặt tổng tiền bằng 50,000 VND (mức tối thiểu). 

```python

def smart_taxi_fare(car_type, start_hour,km, wait_minutes):
    tx=0
    tc=0
    tt=0
    if wait_minutes>5:
        tc=(wait_minutes-5)*2000
    else:
        tc=0

    if car_type == "a":
        if km<=0.8:
            tx=11000
        elif km<=20:
            tx=11000+(km-0.80)*13000
        elif km>20:
            tx=11000+(km-0.80)*13000+(km-20)*10500
    else:
        if km<=0.8:
            tx=12000
        elif km<=20:
            tx=12000+(km-0.80)*14000
        elif km>20:
            tx=12000+(km-0.80)*14000+(km-20)*12000
    if start_hour>=22 or start_hour<6:
        tt=(tx+tc)*0.2+tx+tc
    else:
        tt=tx+tc
    if km>30 and start_hour<22 or start_hour>6:
        tt=tt*0.9
    if tt<50000 and km>5:
        tt=50000
    return tt
car_type,start_hour,km,wait_minutes= input("Nhập loại xe (a/b), giờ bắt đầu, số km, số phút chờ: ").split() 
start_hour=int(start_hour)
km=float(km)  
wait_minutes=int(wait_minutes)
if smart_taxi_fare(car_type, start_hour,km, wait_minutes):
    print("Cước phí taxi thông minh là:", smart_taxi_fare(car_type, start_hour,km, wait_minutes))
```

```python
#Cách 2
def solve_quadratic(a, b, c):
    if a == 0:
        if b != 0:
            return [-c / b]
        else:
            return []
    else:
        d = b**2 - 4*a*c
        if d < 0:
            return []
        elif d == 0:
            return [-b / (2*a)]
        else:
            # TỰ TÍNH CĂN BẬC 2 BẰNG MŨ 0.5
            can_d = d ** 0.5
            nghiem_1 = (-b - can_d) / (2*a)
            nghiem_2 = (-b + can_d) / (2*a)
            
            return sorted([nghiem_1, nghiem_2])

# --- NHẬP VÀ GỌI HÀM ---
a, b, c = map(float, input().split())
print(solve_quadratic(a, b, c))
```

*16.Viết hàm trả về góc ∠ABC theo đơn vị độ khi biết tọa độ các điểm A(x1,y1), B(x2,y2), C(x3,y3).

```python
import math

def tg(x1, y1, x2, y2, x3, y3):
    t = (x1-x2)*(x3-x2) + (y1-y2)*(y3-y2)
    m = (((x1-x2)**2 + (y1-y2)**2)**0.5) * (((x3-x2)**2 + (y3-y2)**2)**0.5)
    
    # g đang là Cos của góc
    g = t/m 
    
    # 1. Dùng math.acos() để dịch Cos sang góc (nhưng đơn vị đang là Radian)
    goc_radian = math.acos(g)
    
    # 2. Dùng math.degrees() để đổi từ Radian sang Độ chuẩn xác
    goc_do = math.degrees(goc_radian)
    
    return goc_do

# --- PHẦN NHẬP DỮ LIỆU CỦA BẠN GIỮ NGUYÊN ---
x1, y1, x2, y2, x3, y3 = map(float, input("Nhập tọa độ x1 y1 x2 y2 x3 y3: ").split())
print(tg(x1, y1, x2, y2, x3, y3))
```

```python
def point_in_quadrilateral(quad, p):
    x, y = p
    so_lan_cat = 0
    
    # Tứ giác có 4 đỉnh, ta sẽ duyệt qua 4 cạnh của nó
    # Cạnh 1: Đỉnh 0 -> 1 | Cạnh 2: Đỉnh 1 -> 2 | Cạnh 3: Đỉnh 2 -> 3 | Cạnh 4: Đỉnh 3 -> 0
    for i in range(4):
        # Điểm đầu của cạnh
        x1, y1 = quad[i]
        # Điểm cuối của cạnh (Dùng % 4 để đỉnh cuối nối ngược lại đỉnh đầu)
        x2, y2 = quad[(i + 1) % 4]
        
        # --- BƯỚC 1: KIỂM TRA ĐIỂM CÓ ĐẠP LÊN CẠNH (BIÊN) KHÔNG ---
        # 1. Điểm p phải nằm trong vùng giới hạn của đoạn thẳng
        trong_gioi_han = (min(x1, x2) <= x <= max(x1, x2)) and (min(y1, y2) <= y <= max(y1, y2))
        
        # 2. Phải thỏa mãn phương trình đường thẳng (Tính bằng vector)
        nam_tren_duong_thang = (x - x1) * (y2 - y1) == (x2 - x1) * (y - y1)
        
        if trong_gioi_han and nam_tren_duong_thang:
            return True # Đạp lên biên -> Chốt sổ là True luôn!

        # --- BƯỚC 2: PHÓNG TIA LASER SANG PHẢI ---
        # Điều kiện 1: Tia nằm ngang (y) phải đâm xuyên qua trục dọc của cạnh này
        # (Nghĩa là y của điểm p phải nằm kẹp giữa y1 và y2)
        if (y1 > y) != (y2 > y):
            # Tính toán xem tia laser sẽ cắt cạnh này ở tọa độ X là bao nhiêu
            toa_do_x_giao_cat = (x2 - x1) * (y - y1) / (y2 - y1) + x1
            
            # Điều kiện 2: Tia phóng sang phải (x -> dương), 
            # nên điểm cắt phải nằm bên phải điểm p
            if x < toa_do_x_giao_cat:
                so_lan_cat += 1 # Ghi nhận 1 lần cắt!

    # --- KẾT LUẬN ---
    # Nếu đếm ra số lẻ (mod 2 != 0) -> True (Ở trong)
    # Nếu đếm ra số chẵn (mod 2 == 0) -> False (Ở ngoài)
    return so_lan_cat % 2 != 0

# ==================================
# --- KIỂM THỬ XEM HOẠT ĐỘNG KHÔNG ---
# ==================================
chuong = [(0, 0), (0, 4), (4, 4), (4, 0)] # Đây là chuồng hình vuông 4x4

ga_1 = (2, 2) # Đứng giữa chuồng
ga_2 = (5, 5) # Đứng tít ngoài chuồng
ga_3 = (4, 2) # Đứng đạp lên cái dây bên phải

print("Gà 1:", point_in_quadrilateral(chuong, ga_1)) # Ra True
print("Gà 2:", point_in_quadrilateral(chuong, ga_2)) # Ra False
print("Gà 3:", point_in_quadrilateral(chuong, ga_3)) # Ra True
```

(Sthực dấu chấm động (float - floating-point): Cứ hiểu đơn giản nó là số thập phân)

```python
def detect_number_type(s):
    # Lột dấu
    if len(s) > 0 and (s[0] == '+' or s[0] == '-'):
        s = s[1:]
        
    if len(s) == 0:
        return 'neither'

    # SỔ GHI CHÉP: Phải đặt ở NGOÀI vòng lặp
    sdc = 0
    vi_tri_dau_cham = -1

    # VÒNG LẶP: Dùng range() và chỉ làm nhiệm vụ ghi chép
    for i in range(len(s)):
        kt = s[i]
        
        if kt == '.':
            sdc += 1
            vi_tri_dau_cham = i # Lưu lại nó nằm ở vị trí nào
            
        elif not kt.isdigit():
            # Chỉ khi gặp CHỮ LẠ (như 'a', 'b') thì mới được quyền tử hình luôn
            return 'neither'

    # TUYÊN ÁN: Đặt ở tít dưới cùng, bằng cấp với chữ `for`
    if sdc == 0:
        return 'int'
        
    elif sdc == 1:
        # Nếu có 1 dấu chấm, phải check xem nó có nằm ở 2 cái rìa không (như .5 hoặc 5.)
        if vi_tri_dau_cham > 0 and vi_tri_dau_cham < len(s) - 1:
            return 'float'
        else:
            return 'neither'
            
    else:
        # Lớn hơn 1 dấu chấm (ví dụ 1.2.3)
        return 'neither'

s = input()
print(detect_number_type(s))

        
```

```python
def is_perfect_square(n):
    if n<0:
        return False
    if n==0:
        return True
    for i in range(1,int(n**0.5)+1):
        if i**2==n:
            return True
    else:
        return False
        
n=int(input())
print(is_perfect_square(n))
```

chu_cai = chr(65)

```python
def dc(x,n):
    if not x.isalpha():
        return x
    if x.upper():
        mocA=ord('A')
        mocZ=ord('Z')
    else:
        mocA=ord('a')
        mocZ=ord('z')
    sc=mocZ-mocA+1
    vitrimoi=(ord(x)-mocA+n)%sc
    return chr(vitrimoi+mocA)
x=input()
n=int(input())
print(dc(x,n))
```
