Nhập vào số nguyên đại diện cho tháng trong năm nhuận. Viết ct in ra số ngày của tháng đó

```python
n = int(input())
if n==1 or n==3 or n==5 or n==7 or n==8 or n==10 or n==12:
    print(f"Tháng {n} có 31 ngày")
elif n==4 or n==6 or n==9 or n==11:
    print(f"Tháng {n} có 30 ngày")
else:
    print(f"Tháng {n} có 29 ngày")
```

```python
# Hàm kiểm tra số nguyên tố cực nhanh
def is_prime(x):
    if x < 2:
        return False
    for i in range(2, int(x**0.5) + 1):
        if x % i == 0:
            return False
    return True

# Đọc liên tục các dòng test case
while True:
    try:
        a, b = map(int, input().split())
        tong = 0
        
        # Duyệt từ a đến b, nếu là số nguyên tố thì cộng dồn
        for i in range(a, b + 1):
            if is_prime(i):
                tong += i
                
        print(tong)
    except EOFError:
        break  # Máy chấm dừng truyền dữ liệu thì thoát vòng lặp
```

W4A9

```python
n = int(input())
ket_qua = []
i = 1

while i * i <= n:
    s = str(i * i)
    # set(s) sẽ xóa các chữ số trùng nhau. Ví dụ set('144') chỉ còn {'1', '4'}
    if len(set(s)) == len(s):
        ket_qua.append(s)
    i += 1

# In mảng cách nhau bằng dấu cách, nếu mảng rỗng thì in (no number)
print(" ".join(ket_qua) if ket_qua else "(no number)")
```

```python
n = int(input())
max_len = 0
best_x = 1

for i in range(1, n + 1):
    x, length = i, 1
    while x != 1:
        x = x // 2 if x % 2 == 0 else 3 * x + 1
        length += 1
        
    if length > max_len:
        max_len = length
        best_x = i

print(f"{best_x} {max_len}")
```

W4A13

```python
def tong_uoc(n):
    s, i = 0, 1
    while i * i <= n:
        if n % i == 0:
            s += i
            # Không cộng chính nó, và tránh cộng trùng ước chính phương
            if i != 1 and i * i != n:
                s += n // i
        i += 1
    return s

a, b = map(int, input().split())

# Kiểm tra chéo theo đúng định nghĩa
if tong_uoc(a) == b and tong_uoc(b) == a:
    print("true")
else:
    print("false")
```
