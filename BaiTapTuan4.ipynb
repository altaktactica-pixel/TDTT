W4A6 
"[SumPrimesInRange] 
Nhập hai số nguyên dương a, b (a ≤ b). Hãy tính tổng các số nguyên tố trong đoạn 
[a, b]. Mỗi dòng là một bộ test (a b)."

```python
# Hàm kiểm tra số nguyên tố cực nhanh
def is_prime(x):
    if x < 2:
        return False
    for i in range(2, int(x**0.5) + 1):
        if x % i == 0:
            return False
    return True

# Đọc liên tục các dòng test case
while True:
    try:
        a, b = map(int, input().split())
        tong = 0
        
        # Duyệt từ a đến b, nếu là số nguyên tố thì cộng dồn
        for i in range(a, b + 1):
            if is_prime(i):
                tong += i
                
        print(tong)
    except EOFError:
        break  # Máy chấm dừng truyền dữ liệu thì thoát vòng lặp
```

Nhập vào số nguyên dương n (n ≥ 2). Hãy in ra ước số nguyên tố lớn nhất của n." 


```python
n = int(input())
i = 2

# Chỉ lặp đến căn bậc 2 của n để tối ưu tốc độ
while i * i <= n:
    if n % i == 0:
        n //= i  # Nếu chia hết, gọt nhỏ n đi
    else:
        i += 1   # Nếu không chia hết, thử số tiếp theo

# Phần n còn lại cuối cùng chính là ước nguyên tố lớn nhất
print(n)
```

W4A8: ReverseAddPalindrome
Nhập số nguyên dương n. Lặp lại phép toán n = n + reverse(n) cho đến khi n là số 
đối xứng (palindrome). In ra số bước và giá trị palindrome thu được. 

```python
n = int(input())
dem = 0

# Kiểm tra xem chuỗi n có giống với chuỗi n đảo ngược không
while str(n) != str(n)[::-1]:
    # Ép chuỗi đảo ngược về int rồi cộng dồn
    n += int(str(n)[::-1])
    dem += 1

print(dem, n)
```

W4A9
[PerfectSquaresDistinctDigits] 
Nhập vào số nguyên dương n. In ra tất cả các số chính phương ≤ n có các chữ số 
đôi một khác nhau (in theo thứ tự tăng dần, cùng hàng, cách nhau bởi một dấu 
cách). Nếu không có, in ra (no number).

```python
n = int(input())
ket_qua = []
i = 1

while i * i <= n:
    s = str(i * i)
    # set(s) sẽ xóa các chữ số trùng nhau. Ví dụ set('144') chỉ còn {'1', '4'}
    if len(set(s)) == len(s):
        ket_qua.append(s)
    i += 1

# In mảng cách nhau bằng dấu cách, nếu mảng rỗng thì in (no number)
print(" ".join(ket_qua) if ket_qua else "(no number)")
```

W4A10
"[LongestCollatzUpToN] 
Nhập vào số nguyên dương n. Tìm số x trong [1, n] có độ dài dãy Collatz lớn nhất 
và in ra x cùng độ dài dãy tương ứng (x L). Nếu có nhiều số cùng độ dài, chọn số 
nhỏ nhất." 

```python
n = int(input())
max_len = 0
best_x = 1

for i in range(1, n + 1):
    x, length = i, 1
    while x != 1:
        x = x // 2 if x % 2 == 0 else 3 * x + 1
        length += 1
        
    if length > max_len:
        max_len = length
        best_x = i

print(f"{best_x} {max_len}")
```

[EvenDivisors] 
Viết chương trình đếm số lượng ước số chẵn của số nguyên dương n với n < 10^6. 

```python
n = int(input())
dem = 0
i = 1

while i * i <= n:
    if n % i == 0:
        if i % 2 == 0: 
            dem += 1
        # Tránh đếm trùng khi i và n//i bằng nhau (ví dụ n=16, i=4)
        if (n // i) != i and (n // i) % 2 == 0: 
            dem += 1
    i += 1

print(dem)
```

Hai số nguyên dương tạo thành một cặp số thân thiết khi chúng tuân theo quy luật: 
Số này bằng tổng tất cả các ước của số kia (trừ chính nó) và ngược lại. 
Ví dụ 220 và 284 là một cặp số thân thiết 
220 ngoài chính nó ra, nó còn có 11 ước số khác là: 1, 2, 4, 5, 10, 11, 20, 44, 55, 
110. Tổng của 11 ước này cho chúng ta 284 
284 ngoài bản thân nó, còn có 5 ước số khác là: 1, 2, 4, 71, 142. Tổng của chúng 
vừa đúng bằng 220 
Nhập hai số từ bàn phím và kiểm tra xem chúng có phải là cặp số thân thiết không. 
Nếu có, in ra màn hình "true", ngược lại, in ra "false" 

```python
def tong_uoc(n):
    s, i = 0, 1
    while i * i <= n:
        if n % i == 0:
            s += i
            # Không cộng chính nó, và tránh cộng trùng ước chính phương
            if i != 1 and i * i != n:
                s += n // i
        i += 1
    return s

a, b = map(int, input().split())

# Kiểm tra chéo theo đúng định nghĩa
if tong_uoc(a) == b and tong_uoc(b) == a:
    print("true")
else:
    print("false")
```
