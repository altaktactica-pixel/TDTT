```python

c = input("Nhập một kí tự chữ cái thường (a-z): ")


ma_unicode = ord(c)
print(f"Mã Unicode của '{c}' là: {ma_unicode}")


chu_hoa = chr(ma_unicode - 32)
print(f"Kí tự hoa tương ứng là: {chu_hoa}")
```
