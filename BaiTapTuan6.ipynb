W6A4: Nhập một dòng gồm các cặp key:value cách nhau bởi dấu cách. Tạo dictionary ánh xạ mỗi key -> list các value theo thứ tự xuất hiện. Nếu key lặp lại, thêm vào list của key đó 

```python
# Bước 1: Nhập dữ liệu và tách thành các cặp key:value
# Ví dụ: "a:1 b:2 a:3" -> ["a:1", "b:2", "a:3"]
du_lieu_vao = input().split()

# Bước 2: Khởi tạo dictionary rỗng
ket_qua = {}

# Bước 3: Duyệt qua từng cặp
for cap in du_lieu_vao:
    # Tách key và value dựa trên dấu ":"
    key, value = cap.split(':')
    
    # KIỂM TRA: Nếu key đã có trong từ điển rồi
    if key in ket_qua:
        # Lấy cái list cũ ra và append thêm giá trị mới vào
        ket_qua[key].append(value)
    else:
        # Nếu chưa có, tạo một khóa mới với giá trị là một List chứa value
        ket_qua[key] = [value]

# Bước 4: In kết quả
print(ket_qua)
```

W4A5: Nhập một list số nguyên, In ra dictionary {'positives': ..., 'negatives': ..., 
'zeros': ...} (đếm số >0, <0, =0) 

```python
#Nếu bắt phân loại số trong lst
d=list(map(int,input().split()))
kq={'postives':[],'negatives':[],'zeros':[]}
for i in d:
    if i>0:
        kq['postives'].append(i)
    elif i<0:
        kq['negatives'].append(i)
    else:
        kq['zeros'].append(i)
print(kq)
        
```

```python
#Theo Đề
danh_sach = list(map(int, input().split()))

# 2. Tạo sẵn cuốn sổ tay (Dictionary) với 3 trang, ban đầu số lượng đều là 0
ket_qua = {
    'positives': 0,
    'negatives': 0,
    'zeros': 0
}

# 3. Quét từng số trong danh sách để phân loại
for so in danh_sach:
    if so > 0:
        ket_qua['positives']+=1  # Ghi thêm 1 vạch vào trang positives
    elif so < 0:
        ket_qua['negatives'] += 1  # Ghi thêm 1 vạch vào trang negatives
    else:
        ket_qua['zeros'] += 1      # Ghi thêm 1 vạch vào trang zeros

# 4. In cuốn sổ ra màn hình
print(ket_qua)
```

*W6A9: Viết chương trình nhập vào hai dictionary. Sau đó gộp lại thành một dictionary mới trong đó nếu một key xuất hiện trong cả hai dict, giá trị sẽ được cộng lại, nếu chỉ có trong một dict thì giữ nguyên. In ra dictionary mới theo thứ tự từ điển tăng dần của key. 

```python

def nhap_dict():
    chuoi_nhap = input().split()
    d = {}
    for item in chuoi_nhap:
        key, value = item.split(':')
        d[key] = int(value) # Nhớ ép kiểu Value thành số nguyên để cộng
    return d

dict1 = nhap_dict()
dict2 = nhap_dict()

dict_gop = dict1.copy()


for k, v in dict2.items():
    if k in dict_gop:
        dict_gop[k] += v  
    else:
        dict_gop[k] = v   

# Lệnh sorted() sẽ lôi các cặp ra sắp xếp theo Key, trả về một List các Tuple
# Ta dùng lệnh dict() để ép cái List đó trở lại thành Dictionary
dict_ket_qua = dict(sorted(dict_gop.items()))

print(dict_ket_qua)
```

W6A10: Viết chương trình nhập vào một danh sách các số nguyên, và số nguyên dương k. In ra list gồm các cặp số khác nhau có tổng bằng k. Mỗi cặp là một tuple, hai số trong tuple có thể giống nhau, thứ tự xuất hiện trong danh sách của số thứ nhất nhỏ hơn thứ tự của số thứ hai. Các tuple trong list được sắp xếp theo thứ tự tăng dần của phần tử đầu tiên trong tuple. 

```python
ds=list(map(int,input().split()))
k=int(input())
caccapso=set()
for i in range(len(ds)):
    for j in range (i+1,len(ds)):
        if ds[i]+ds[j]==k:
            capmoi=(ds[i],ds[j])
            caccapso.add(capmoi)
kq=list(caccapso)
kq.sort()
print(kq)
```

W6A13: [Đảo ngược key và value của dictionary] 
Nhập vào một dictionary có value là duy nhất (không trùng nhau). Hãy tạo một dictionary mới bằng cách đảo ngược key và value của dictionary ban đầu. In ra dictionary mới.

```python
ds=input().split()
d={}
for item in ds:
    k,v=item.split(":")
    d[k]=v
d_new={}
for k,v in d.items():
    d_new[v]=k
print(d_new)
```

W6A15: [Lọc dictionary theo giá trị] 
Nhập vào một dictionary có key là chuỗi và value là số nguyên, và một số nguyên k ở dòng tiếp theo. Hãy tạo một dictionary mới chỉ chứa các cặp key-value từ dictionary ban đầu mà có value lớn hơn k. In ra dictionary mới.

```python
ds=input().split()
d={}
for i in ds:
    key,v=i.split(":")
    d[key]=int(v)
k=int(input())
d_new={}
for keys,values in d.items():
    if values>k:
        d_new[keys]=values
print(d_new)
```
