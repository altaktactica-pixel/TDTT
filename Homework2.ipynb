Bài 3:  Tính tam giác: Nhâp vào ba số a,b,c bất kì. Kiểmtra xem nó có thể làđộ dài ba cạnh hay không, nếu không thì in ra màn hình 'Khong la ba canh cua mot tam giac'. Ngược lại, kiểm tra xem nó là tam giác gì: Đều, Cân, Vuông, hay Tam giác thường. Tính diện tích, chu vi của tam giác và in ra mànhình.

```python
import math
a,b,c=map(float,input().split())
if a+b>c and a+c>b and b+c>a:
    if a==b and b==c:
        print('Tam giác đều')
    elif a==b or b==c or a==c:
        print('Tam giác cân')
    elif a**2+b**2==c**2 or a**2+c**2==b**2 or b**2+c**2==a**2:
        print('Tam giác vuông')    
    else:
        print('Tam giác thường')
else:
    print('Không là 3 cạnh của một tam giác')
```

Bài 5:  Viết chương trình giải phương trình bậc 2 (ax2 + bx + c = 0)

```python
import math

a, b, c = map(float, input().split())

if a == 0:
    # Nếu a = 0, bài toán quay về phương trình bậc 1: bx + c = 0
    if b == 0:
        if c == 0:
            print("Phương trình vô số nghiệm")
        else:
            print("Phương trình vô nghiệm")
    else:
        print(f"{-c / b:.2f}")
else:
    delta = b**2 - 4*a*c
    
    if delta < 0:
        print("Phương trình vô nghiệm")
    elif delta == 0:
        x = -b / (2*a)
        print(f"Phương trình có nghiệm kép: x = {x:.2f}")
    else:
        x1 = (-b + math.sqrt(delta)) / (2*a)
        x2 = (-b - math.sqrt(delta)) / (2*a)
        print(f"Phương trình có hai nghiệm phân biệt: x1 = {x1:.2f}, x2 = {x2:.2f}")
```

Bài 8: Giải hệ phương trình tuyến tính:
ax  by  m
cx  dy  n

```python
# Cramer
a, b, m = map(float, input().split())
c, d, n = map(float, input().split())

# Tính các định thức
D = a * d - b * c
Dx = m * d - b * n
Dy = a * n - m * c

if D == 0:
    if Dx == 0 and Dy == 0:
        print("Vô số nghiệm")
    else:
        print("Vô nghiệm")
else:
    x = Dx / D
    y = Dy / D
    # In ra x và y (có thể tuỳ chỉnh làm tròn theo yêu cầu hệ thống)
    print(f"x={x:.2f}, y={y:.2f}")
```

Bài 9: Nhập vào thời gian 1 công việc nào đó là x giây. Hãy chuyển đổi và viết ra màn hình số thời gian trên dưới dạng bằng bao nhiêu giờ, bao nhiêu phút, bao nhiêu giây. 

```python
x = int(input().strip())

gio = x // 3600
phut = (x % 3600) // 60 
giay = x % 60
print(f"{gio} giờ {phut} phút {giay} giây")
```

Bài 10:  Nhập vào tâm và bán kính của một đường tròn. Sau đó nhập vàomộtđiểm A(x, y) bất kì và kiểm tra xem nó có thuộc đường tròn hay không?

```python
import math
xc, yc, r = map(float, input().split())
xa, ya = map(float, input().split())
khoang_cach_b2 = (xa - xc)**2 + (ya - yc)**2
ban_kinh_b2 = r**2

# Dùng <= nếu tính cả bên trong, dùng == nếu chỉ tính trên viền ngoài
if khoang_cach_b2 <= ban_kinh_b2:
    print("Thuộc đường tròn")
else:
    print("Không thuộc đường tròn")
```
