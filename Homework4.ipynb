W4A20:  Nhập vào một số nguyên dương n, kiểm tra xem n có phải là số dạng 2^k hay không 

```python
while True:
    n=int(input())
    if n>0:
        break
while n%2==0:
    n=n//2
if n==1:
    print(f"Số nhập vào là dạng 2^k")
else:
    print(f"Số nhập vào không phải là dạng 2^k")
```

W4A25: Nhập vào một dãy số nguyên, ngưng nhập khi người dùng nhập -1. 
Sau khi nhập xong, in số lớn nhất, số nhỏ nhất trong những số vừa nhập

```python
mang_so = []
# BƯỚC 1: Nhập dữ liệu và lưu vào mảng
while True:
    n = int(input())
    if n == -1:
        break
    mang_so.append(n)

# BƯỚC 2: Dùng chỉ số (index) và vòng lặp để tìm max, min
so_luong_phan_tu = len(mang_so)

if so_luong_phan_tu == 0:
    print("Bạn chưa nhập số nào ngoài -1.")
else:
    # Lấy phần tử ở chỉ số 0 làm mốc mặc định
    lon_nhat = mang_so[0]
    nho_nhat = mang_so[0]
    
    # Dùng vòng lặp for chạy biến i (đóng vai trò là chỉ số) từ 1 đến hết mảng
    for i in range(1, so_luong_phan_tu):
        # mang_so[i] chính là giá trị của mảng tại vị trí thứ i
        if mang_so[i] > lon_nhat:
            lon_nhat = mang_so[i]
            
        if mang_so[i] < nho_nhat:
            nho_nhat = mang_so[i]
            
    print("Số lớn nhất là:", lon_nhat)
    print("Số nhỏ nhất là:", nho_nhat)
 
```

W4A26: [Fibonaci1] 
Dãy số fibonacci là dãy số được định nghĩa như sau: 1, 1, 2, 3, 5, 8, 13,... với số kế tiếp sẽ 
bằng tổng hai số trước đó 
Nhập vào số nguyên dương A, hãy tìm số trong dãy số fibonacci lớn nhất nhưng không vượt 
quá A 

```python
while True:
    A=int(input())
    if A>0:
        break
f1 = 1
f2 = 1
ket_qua = 1
while f1 <= A:
    ket_qua = f1
    f_tiep_theo = f1 + f2
    f1 = f2
    f2 = f_tiep_theo
print(ket_qua)
```

W4A27: Nhập vào một chuỗi, hãy đếm xem trong chuỗi có bao nhiêu từ (quy định là chuỗi không có 
ký tự đặc biệt, không số, không có dấu câu, chỉ có ký tự chữ và khoảng trắng) 
Dùng cách: for trên index; KHÔNG dùng hàm split(), len() 

```python
chuoi = input()

# --- BƯỚC 1: TÌM CHIỀU DÀI CHUỖI (Thay thế hàm len) ---
do_dai = 0
for ky_tu in chuoi:
    do_dai += 1


# --- BƯỚC 2: ĐẾM TỪ BẰNG INDEX ---
dem_tu = 0

# Cho ngón tay i chạy từ vị trí 0 đến vị trí cuối cùng của chuỗi
for i in range(do_dai):
    
    # Kiểm tra: Nếu ký tự hiện tại đang trỏ vào là một chữ cái (không phải khoảng trắng)
    if chuoi[i] != " ":
        
        # Kiểm tra tiếp: Nó có phải là chữ cái bắt đầu của một từ không?
        # Đúng nếu nó nằm ở đầu chuỗi (i == 0) 
        # HOẶC ký tự ngay trước nó (i - 1) là một khoảng trắng
        if i == 0 or chuoi[i - 1] == " ":
            dem_tu += 1  # Bấm bộ đếm lên 1!

print(dem_tu)
```

W4A28: Nhập vào một chuỗi, hãy in từ đầu tiên trong chuỗi 
Dùng cách: for trên index; KHÔNG dùng hàm split()

```python
chuoi = input()

# Tìm chiều dài chuỗi (nếu bài cấm cả hàm len() thì dùng cách đếm giống bài trước)
do_dai = 0
for ky_tu in chuoi:
    do_dai += 1

# Chuẩn bị cái giỏ trống
tu_dau_tien = ""

for i in range(do_dai):
    
    # Nếu ký tự hiện tại KHÔNG phải là khoảng trắng -> Nó là chữ cái
    if chuoi[i] != " ":
        # Nhặt chữ đó ghép vào giỏ (dấu + dùng để nối chuỗi)
        tu_dau_tien += chuoi[i]
        
    # Nếu ký tự hiện tại LÀ khoảng trắng
    else:
        # Kiểm tra xem giỏ đã có đồ chưa?
        # Nếu giỏ khác rỗng (nghĩa là đã nhặt được ít nhất 1 chữ)
        if tu_dau_tien != "":
            break  # Vỡ vòng lặp ngay lập tức vì đã lấy xong từ đầu tiên!

print(tu_dau_tien)
```

W4A29: Cho một chuỗi chỉ gồm các ký tự chữ cái thường và chữ cái hoa, hãy đổi ký tự hoa thành 
thường và thường thành hoa. 

```python
chuoi = input()
ket_qua = ""  

for ky_tu in chuoi:
    if ky_tu.islower():
        ket_qua += ky_tu.upper()
    elif ky_tu.isupper():
        ket_qua += ky_tu.lower()
    else:
        ket_qua += ky_tu
print(ket_qua)
```

W4A30: Nhập vào một chuỗi có dạng 3 số nguyên, mỗi số nguyên cách nhau một dấu phẩy, hãy tính

W4A31: Nhập vào một chuỗi, hãy đếm xem trong chuỗi có bao nhiêu ký tự in hoa, bao nhiêu ký tự in 
thường, bao nhiêu ký tự số

```python
chuoi = input()
dem_hoa = 0
dem_thuong = 0
dem_so = 0
for ky_tu in chuoi:
    if ky_tu.isupper():
        dem_hoa += 1
    elif ky_tu.islower():
        dem_thuong += 1
    elif ky_tu.isdigit():
        dem_so += 1
print(dem_hoa)
print(dem_thuong)
print(dem_so)
```
