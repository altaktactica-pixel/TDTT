W7A1:Viết hàm tìm kiếm nhị phân trên một danh sách đã được sắp xếp tăng dần. Trả về chỉ số của phần tử hoặc -1 nếu không tìm thấy. mẫu hàm: def binary_search(arr, left, right, target):

```python
def binary_search(arr, left, right, target):
    # Điều kiện dừng: Nếu biên trái vượt qua biên phải, nghĩa là đã tìm hết mảng mà không thấy
    if left > right:
        return -1
    
    # Tìm vị trí đứng ở chính giữa mảng
    mid = (left + right) // 2
    
    # Trường hợp 1: Trúng phóc! Phần tử ở giữa chính là số cần tìm
    if arr[mid] == target:
        return mid
        
    # Trường hợp 2: Số ở giữa LỚN HƠN số cần tìm. 
    # Vì mảng đã sắp xếp tăng dần, số cần tìm chắc chắn nằm ở nửa BÊN TRÁI.
    # Ta vứt bỏ nửa bên phải, dời biên phải (right) về sát bên trái của mid.
    elif arr[mid] > target:
        return binary_search(arr, left, mid - 1, target)
        
    # Trường hợp 3: Số ở giữa NHỎ HƠN số cần tìm.
    # Tương tự, số cần tìm chắc chắn nằm ở nửa BÊN PHẢI.
    # Ta vứt bỏ nửa bên trái, dời biên trái (left) sang ngay bên phải của mid.
    else:
        return binary_search(arr, mid + 1, right, target)


danh_sach = [2, 5, 8, 12, 16, 23, 38, 56, 72, 91] # Danh sách BẮT BUỘC phải sắp xếp tăng dần
so_can_tim = 23

# Lần gọi đầu tiên: left là 0 (đầu mảng), right là len - 1 (cuối mảng)
vi_tri = binary_search(danh_sach, 0, len(danh_sach) - 1, so_can_tim)

print(f"Danh sách: {danh_sach}")
if vi_tri != -1:
    print(f"Đã tìm thấy số {so_can_tim} tại vị trí (index): {vi_tri}")
else:
    print(f"Không tìm thấy số {so_can_tim} trong danh sách.")
```

W7A3: Viết hàm thực hiện thuật toán bubble sort, Selection Sort để sắp xếp danh sách theo thứ tự tăng dần. Sắp xếp nổi bọt (Bubble Sort);Sắp xếp chọn (Selection Sort).

```python
def bubble_sort(arr):
    n=len(arr)
    for i in range(n):
        swapped=True
        for j in range(0,n-i-1):
            if arr[j]>arr[j+1]:
                arr[j],arr[j+1]=arr[j+1],arr[j]
                swapped=False
    return arr
arr=list(map(int,input().split()))
print(bubble_sort(arr))
```

```python
def selection_sort(arr):
   n=len(arr)
   for i in range(n):
      min_idx=i
      for j in range (i+1,n):
         if arr[j]<arr[min_idx]:
            min_idx=j
            arr[i],arr[min_idx]=arr[min_idx],arr[i]
   return arr
arr=list(map(int,input().split()))
print(selection_sort(arr))
```

W7A4: Viết hàm tìm phần tử có số lần xuất hiện nhiều nhất trong danh sách. Nếu có nhiều phần tử cùng số lần, trả về phần tử xuất hiện trước.

```python
def s(lst):
    if len(lst)==0:
        return None
    cnt={}
    for so in lst:
        if so in cnt:
            cnt[so]+=1
        else:
            cnt[so]=1
    solanmax=0
    kq=0
    for so,solan in cnt.items():
        if solan>solanmax:
            solanmax=solan
            kq=so
    return kq
lst=list(map(int,input().split()))
print(s(lst))
```

W7A6: Bạn được cung cấp một danh sách các danh sách lồng nhau, vd [[1, 2, 3], [4, 5], [6, 7, 8, 9]], hãy làm phẳng danh sách

```python
def lamphang(lst):
    kq=[]
    for i in range(len(lst)):
        mc=lst[i]
        if i%2!=0:
            kq.extend(mc[::-1])
        else:
            kq.extend(mc)
    return kq
n=input()
lst=eval(n)
print(lamphang(lst))
```

W7A7: Cho một danh sách các số nguyên nums, hãy tìm độ dài của dãy con tăng nghiêm ngặt dài nhất (không nhất thiết phải liền kề).

```python
def tim_chieu_dai_LIS(nums):
    # Nếu danh sách rỗng, chiều dài là 0
    if not nums:
        return 0
        
    n = len(nums)
    # Khởi tạo "sổ tay" dp: ban đầu mỗi số tự tạo thành 1 dãy có độ dài 1
    dp = [1] * n
    
    # Duyệt qua từng số trong mảng
    for i in range(1, n):
        # Nhìn lại tất cả các số đứng trước nó
        for j in range(0, i):
            # Nếu số đứng trước NHỎ HƠN số hiện tại (thỏa mãn dãy tăng)
            if nums[j] < nums[i]:
                # Cập nhật sổ tay: Lấy kỷ lục lớn nhất
                dp[i] = max(dp[i], dp[j] + 1)
                
    # Đáp án là con số lớn nhất được ghi trong sổ tay
    return max(dp)

chuoi_nhap = input().strip()

danh_sach = eval(chuoi_nhap)
    
print(tim_chieu_dai_LIS(danh_sach))
    
```

W7A8: Bạn được cung cấp: một danh sách các khoảng thời gian interval = [[l1, r1], [l2, r2], ...] và một danh sách các truy vấn queries = [q1, q2, ...] Với mỗi truy vấn q, hãy tìm độ dài nhỏ nhất của khoảng thời gian [li, ri] sao cho li ≤ q ≤ ri. Nếu không tồn tại khoảng thời gian nào như vậy, hãy trả về -1 cho truy vấn đó. solve in < O(m.n), hint: binary search in the queries list

```python
def solve():
    try:
        intervals = eval(input())
        queries = eval(input())
    except EOFError:
        return

    # Sắp xếp các khoảng từ NGẮN đến DÀI
    intervals.sort(key=lambda x: x[1] - x[0] + 1)

    results = []

    # Duyệt qua từng câu hỏi (truy vấn)
    for q in queries:
        
        # 1. GÁN MẶC ĐỊNH LÀ KHÔNG TÌM THẤY
        dap_an = -1 
        
        # 2. Bắt đầu đi tìm
        for l, r in intervals:
            if l <= q <= r:
                # NẾU TÌM THẤY: Cập nhật lại đáp án bằng độ dài khoảng
                dap_an = r - l + 1
                
                # Tìm thấy rồi thì DỪNG LẠI luôn, không tìm thêm nữa
                break 
                
        # 3. Ghi kết quả vào sổ
        # Nếu vòng lặp trên chạy xong mà không có break (không tìm thấy)
        # thì dap_an vẫn giữ nguyên là -1. 
        # Nếu có break, dap_an đã được đổi thành độ dài ngắn nhất.
        results.append(dap_an)
            
    print(results)

solve()
```

W7A9: Cho một danh sách các số nums, với mỗi phần tử nums[i], hãy đếm xem có bao nhiêu phần tử bên phải nhỏ hơn nums[i]. Trả về danh sách các số đếm. Efficient solution: using heap or binary search tree

```python
def solve():
    try:
        # Đọc danh sách đầu vào (Dùng eval để xử lý chuỗi [5, 2, 6, 1])
        nums = eval(input())
    except EOFError:
        return

    n = len(nums)
    results = [0] * n
    
    # Danh sách chứa các số đã duyệt (luôn được giữ ở trạng thái sắp xếp tăng dần)
    sorted_right = []
    
    # 1. Duyệt ngược từ cuối mảng lên đầu
    for i in range(n - 1, -1, -1):
        x = nums[i] # Số đang xét
        
        # 2. BINARY SEARCH THỦ CÔNG
        # Mục đích: Tìm vị trí (pos) để chèn số x vào sorted_right sao cho vẫn tăng dần
        left, right = 0, len(sorted_right) - 1
        pos = len(sorted_right) # Mặc định là chèn ở cuối
        
        while left <= right:
            mid = (left + right) // 2
            if sorted_right[mid] >= x:
                pos = mid        # Ghi nhận vị trí có thể chèn
                right = mid - 1  # Ép sang trái để tìm vị trí nhỏ nhất
            else:
                left = mid + 1
                
        # 3. Ghi nhận kết quả và Chèn số vào danh sách
        # Vị trí chèn (pos) CHÍNH LÀ số lượng phần tử nhỏ hơn x!
        results[i] = pos
        
        # Chèn x vào vị trí pos của danh sách sorted_right
        sorted_right.insert(pos, x)
        
    print(results)

solve()
```

W7A10: Cho một danh sách các số nguyên nums và một số nguyên m, hãy tìm tổng mảng con lớn nhất theo modulo m. Tức là, với bất kỳ mảng con nums[i:j] nào, hãy tính sum(nums[i:j]) % m và trả về kết quả lớn nhất có thể. Gợi ý: Duy trì tổng tiền tố (prefix sum) mod m và sử dụng tìm kiếm nhị phân để tìm tổng tiền tố nhỏ nhất chỉ lớn hơn tổng hiện tại một chút (để tối đa hóa hiệu số bao quanh).

```python
def solve():
    try:
        nums = eval(input())
        m = int(input())
    except EOFError:
        return

    max_mod_sum = 0
    current_prefix = 0
    
    # Danh sách lưu các tổng tiền tố cũ, luôn được sắp xếp TĂNG DẦN
    sorted_prefixes = []

    for x in nums:
        # 1. Tính tổng cộng dồn hiện tại trên "đồng hồ" m
        current_prefix = (current_prefix + x) % m
        
        # Cập nhật kết quả nếu bản thân tổng từ đầu đến giờ đã là lớn nhất
        if current_prefix > max_mod_sum:
            max_mod_sum = current_prefix
            
        # 2. BINARY SEARCH LẦN 1: 
        # Tìm một số cũ TRONG SỔ lớn hơn current_prefix một chút
        left, right = 0, len(sorted_prefixes) - 1
        pos_larger = -1
        
        while left <= right:
            mid = (left + right) // 2
            if sorted_prefixes[mid] > current_prefix:
                pos_larger = mid
                right = mid - 1 # Tìm thấy rồi nhưng vẫn ép sang trái để kiếm số nhỏ hơn nữa
            else:
                left = mid + 1
                
        # 3. Tính toán "Cú quay xe" thần thánh
        if pos_larger != -1:
            smallest_larger = sorted_prefixes[pos_larger]
            # Công thức bù trừ khi bị âm: (hiện_tại - cũ + m)
            wrap_around_sum = current_prefix - smallest_larger + m
            if wrap_around_sum > max_mod_sum:
                max_mod_sum = wrap_around_sum
                
        # 4. BINARY SEARCH LẦN 2: 
        # Chèn current_prefix vào sổ để dùng cho các bước sau
        left, right = 0, len(sorted_prefixes) - 1
        pos_insert = len(sorted_prefixes)
        
        while left <= right:
            mid = (left + right) // 2
            if sorted_prefixes[mid] >= current_prefix:
                pos_insert = mid
                right = mid - 1
            else:
                left = mid + 1
                
        sorted_prefixes.insert(pos_insert, current_prefix)
        
    print(max_mod_sum)

solve()
```
