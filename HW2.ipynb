Bài 5

```python
import math

a, b, c = map(float, input().split())

if a == 0:
    # Nếu a = 0, bài toán quay về phương trình bậc 1: bx + c = 0
    if b == 0:
        if c == 0:
            print("Phương trình vô số nghiệm")
        else:
            print("Phương trình vô nghiệm")
    else:
        print(f"{-c / b:.2f}")
else:
    delta = b**2 - 4*a*c
    
    if delta < 0:
        print("Phương trình vô nghiệm")
    elif delta == 0:
        x = -b / (2*a)
        print(f"Phương trình có nghiệm kép: x = {x:.2f}")
    else:
        x1 = (-b + math.sqrt(delta)) / (2*a)
        x2 = (-b - math.sqrt(delta)) / (2*a)
        print(f"Phương trình có hai nghiệm phân biệt: x1 = {x1:.2f}, x2 = {x2:.2f}")
```

Bài 8

```python
# Cramer
a, b, m = map(float, input().split())
c, d, n = map(float, input().split())

# Tính các định thức
D = a * d - b * c
Dx = m * d - b * n
Dy = a * n - m * c

if D == 0:
    if Dx == 0 and Dy == 0:
        print("Vô số nghiệm")
    else:
        print("Vô nghiệm")
else:
    x = Dx / D
    y = Dy / D
    # In ra x và y (có thể tuỳ chỉnh làm tròn theo yêu cầu hệ thống)
    print(f"x={x:.2f}, y={y:.2f}")
```

```python
import math
xc, yc, r = map(float, input().split())
xa, ya = map(float, input().split())
khoang_cach_b2 = (xa - xc)**2 + (ya - yc)**2
ban_kinh_b2 = r**2

# Dùng <= nếu tính cả bên trong, dùng == nếu chỉ tính trên viền ngoài
if khoang_cach_b2 <= ban_kinh_b2:
    print("Thuộc đường tròn")
else:
    print("Không thuộc đường tròn")
```
