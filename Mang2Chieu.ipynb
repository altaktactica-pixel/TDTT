Bài 2A: Tính tổng cột thứ k của ma trận nxm số nguyên (n, m vào từ bàn phím)(n là hàng và m là cột)(1<=k<=m>)

```python
n,m,k=map(int,input().split())
matrix=[]
for _ in range(n):
    row=list(map(int,input().split()))
    matrix.append(row)
if 1<=k<=m:
    k-=1
    tong=0
    for i in range(n):
        tong+=matrix[i][k]
    print(tong)
else:
    print("Invalid")
```
