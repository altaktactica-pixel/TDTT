W6A16: Nhập vào mảng n hàng, m cột số nguyên. In lại mảng ra màn hình như ma trận hai chiều, định dạng mỗi phần tử 4 vị trí căn phải 

```python

n, m = map(int, input().split())
ma_tran = []

for _ in range(n):
    dong = list(map(int, input().split()))
    if len(dong)==m:
        ma_tran.append(dong)
    else:
        print("Lỗi")
        break
for dong in ma_tran:
    for so in dong:
        print(f"{so:>4}", end="")
    print()
```

W6A17: Nhập vào mảng vuông 2 chiều n hàng, n cột, in ra đường chéo chính và đường chéo phụ của mảng 

```python
n=int(input())
mt=[]
for _ in range(n):
    dong=list(map(int,input().split()))
    if len(dong)==n:
        mt.append(dong)
    else:
        print("Lỗi")
        break
dcc=[]
dcp=[]

for i in range(n):
    dcc.append(mt[i][i])
    dcp.append(mt[i][n-1-i])

print(dcc)
print(dcp)
```

W6A20: Viết hàm đưa vào một list số nguyên và một số nguyên dương k. Hãy tìm và trả về vị trí của phần tử đầu tiên có giá trị k trong list, nếu không có thì trả về -1

```python
def tkds(ds):
    ds=list(map(int,input().split()))
    k=int(input())
    for i in range(len(ds)):
        if ds[i]==k:
            return i
        else:
            return -1
print(tkds(ds))
```

W6A21:  Nhập một dòng gồm các chuỗi (tách bởi dấu cách). Tạo dictionary ánh xạ chuỗi -> tuple các vị trí (0-based) mà chuỗi đó xuất hiện 

```python
# 1. Nhập và cắt chuỗi thành danh sách các từ
chuoi_nhap = input().split()

# 2. Tạo một cuốn sổ (Dictionary) trống
tu_dien = {}

# 3. Lệnh enumerate() giúp ta lấy ra được cả (Vị trí) lẫn (Từ) cùng một lúc
for vi_tri, tu in enumerate(chuoi_nhap):
    
    # Nếu từ này chưa có trong sổ, tạo ngay một cái "rổ tạm" (List trống) cho nó
    if tu not in tu_dien:
        tu_dien[tu] = []
        
    # Nhét cái vị trí vào rổ của từ đó
    tu_dien[tu].append(vi_tri)

# 4. Đề bắt buộc dùng Tuple, nên ta ép các rổ (List) thành (Tuple)
for tu in tu_dien:
    tu_dien[tu] = tuple(tu_dien[tu])

# 5. In kết quả ra màn hình
print(tu_dien)
```

W6A23: VIết hàm xóa đi các số < x có trong mảng 

```python
def xoa_nguoc(ds, x):
    # Lặp ngược từ vị trí cuối cùng (len-1) về 0 (đến -1 để nó dừng ở 0), bước nhảy -1
    for i in range(len(ds) - 1, -1, -1):
        if ds[i] < x:
            ds.pop(i) # Xóa thoải mái không sợ bỏ sót
    return ds # Xóa xong hết vòng lặp mới return

# --- Chạy thử ---
danh_sach = list(map(int, input("Nhập mảng: ").split()))
dieu_kien = int(input("Nhập x: "))
print(xoa_nguoc(danh_sach, dieu_kien))
```

W6A24: Viết hàm đẩy tất cả các số 0 về đầu (về cuối), các số khác vẫn đúng vị trí trước sau: 0 2 3 2 4 0 0 0 9 => 0 0 0 2 3 2 4 9 

```python
def dc(ds):
    c0=[]
    kc0=[]
    for so in ds:
        if so==0:
            c0.append(so)
        else:
            kc0.append(so)
    return c0+kc0
ds=list(map(int,input().split()))
print(dc(ds))
```

W6A26: Tính tổng các phần tử cực đại địa phương trong một danh sách, không tính 2 phần tử đầu và cuối

W6A28: Cho một dãy các số nguyên, kiểm tra xem nó là dãy tăng thực sự hay giảm thực sự, hay vô danh (không tăng, không giảm) 

```python
ds=list(map(int,input().split()))
if ds==sorted(ds) and len(set(ds))==len(ds):
    print("Là dãy tăng")
elif ds==sorted(ds,reverse=True) and len(set(ds))==len(ds):
    print("Là dãy giảm")
else:
    print("Là dãy vô danh")
```
