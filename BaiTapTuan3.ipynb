W3A3 Kiểm tra 1 số có phải lũy thừa của 2 không
Gợi ý: với lũy thừa của 2, n & (n-1) == 0
 VD: 16 = 00010000 (1*2^4 số 1 ở vị trí thứ 5 từ bên phải sang trái)
     15 = 0000...1

```python
n = int(input())
if n > 0 and (n & (n - 1)) == 0:
    print("n là lũy thừa của 2")
else:
    print("n không phải là lũy thừa của 2")
```

W3A4: Viêt chương trình chia một số m cho n, kết quả thu được làm tròn xuống
LÀM TRÒN LÊN
VD 2/3 = 0.66666
Lấy đến 3 chữ số => nhân 10^3 =>0.667
Cho số thực n và một số nguyên k, in ra số thực n lấy đến k chữ số phần thập phân
0.666000 nếu k=3 => 0.666

```python
# Nhập hai số m và n trên cùng một dòng, cách nhau bởi dấu cách
m, n = map(int, input().split())

# Toán tử // sẽ tự động chia và làm tròn xuống
ket_qua = m // n

# In kết quả
print(ket_qua)
```

W3A5: Viêt chương trình chia một số m cho n, kết quả thu được làm tròn lên

```python
m, n = map(int, input().split())

# Cứ chia lấy phần nguyên trước
ket_qua = m // n

# Nếu chia có dư (tức là phần dư khác 0), ta cộng thêm 1
if m % n != 0:
    ket_qua += 1

print(ket_qua)
```

W3A11: [Phân loại tam giác]
Nhập vào 3 số nguyên dương a,b,c là độ dài 3 cạnh của một tam giác. Kiểm
tra 3 số đã nhập có tạo thành một tam giác hợp lệ hay không. Nếu không in
ra "Không phải tam giác". Nếu có, phân loại tam giác đó là "Tam giác đều",
"Tam giác cân", hay "Tam giác thường".


```python
a,b,c = map(int, input().split())
if a+b>c and a+c>b and b+c>a:
    if a==b==c:
        print("Tam giác đều")
    elif a==b or a==c or b==c:
        print("Tam giác cân")
    else:
        print("Tam giác thường")
else:
    print("Không phải tam giác")
```

nghiệm" hoặc "Vô số nghiệm" trong trường hợp tương ứng.
