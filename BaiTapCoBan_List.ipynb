### Bài 2b: Tìm số nhỏ nhất của 1 list và vị trí của nó, nếu có nhiều số nhỏ nhất bằng nhau thì lấy chỉ số đầu tiên (nhỏ nhất)
10 0 0 12 34 5 6 7  8 89 
0 1

```python
def min_and_pos(lst):
    if lst:
        min_value=lst[0]
        min_pos=0
        for i,x in enumerate(lst,1):
            if x <min_value:
                min_value=x
                min_pos=i
    return min_value, min_pos
assert min_and_pos([1,2,4,5,6,8]) == (1,1)   
assert min_and_pos([]) == None
print("Pass all tests")
```

### Bài 3: Tìm giá trị trung bình của 1 list

```python
def A_list(lst):
    if lst:
        lst_positive=[]
        for x in lst:
            if x>0:
                lst_positive.append(x)
        if not lst_positive:
            return "invalid"
        t=0
        for x in lst_positive:
            t+=x
        return round(t/len(lst_positive),2)
    return "invalid"
assert A_list([2,2,2,-10,-10]) == 2
assert A_list([-2,-1,-20,-343]) == "invalid"
assert A_list([]) == "invalid"
print("Accepted")
```

### Bài 5: Đưa tất cả các số bằng 0 trong list về đầu của list. Các số khác không thay đổi thứ tự trước sau
3 4 6 7 1 9 0 0 23 0 0 34 67 89
0 0 0 0 3 4 6 7 1 9 23 34 67 89

```python
def move_zero_to_first(lst):
    if lst:
        for i,x in enumerate(lst):
            if x ==0:
                lst.pop(i)
                lst.insert(0,x)
        return lst
    return None
assert move_zero_to_first([3,4,0,0,23,0,0,34,67,89]) == [0,0,0,0,3,4,23,34,67,89]
print("Accepted")
```

```python
def move_zero_to_first(lst):
    if lst:
        lst_zero=[]
        lst_nonzero=[]
        for x in lst:
            if x==0:
                lst_zero.append(x)
            else:
                lst_nonzero.append(x)

        return lst_zero +lst_nonzero
    return None
assert move_zero_to_first([3,4,0,0,23,0,0,34,67,89]) == [0,0,0,0,3,4,23,34,67,89]
print("Accepted")
```

### Bài 6: Xóa tất cả các số giống nhau trong list, Các số khác giữ nguyên vị trí trước sau

```python
def unique_arr(lst):
    if lst:
        ans=[]
        for x in lst:
            if x not in ans:
                ans.append(x)
        return ans
    return None
assert unique_arr([1,2,3,4,1,2,3,4])
print("Accepted")
```

### Bài 7: Kiểm tra một list số nguyên có phải là dãy tăng, dãy giảm hay vô danh
1 3 5 6 7 8 9: Tang
9 8 7 6 5 4 0: Giam
1 2 3 7 1 0 2: Vo danh
1 2 2 4 5 7: Vo danh

```python

ds = list(map(int, input().split()))

co_tang = True
co_giam = True

for i in range(len(ds) - 1):
    
    if ds[i] >= ds[i + 1]:
        co_tang = False
    if ds[i] <= ds[i + 1]:
        co_giam = False

if co_tang:
    print("Tang")
elif co_giam:
    print("Giam")
else:
    print("Vo danh")
```

### Bài 8: Kiểm tra một list số nguyên có phải là ziczac khong
1 3 2 5 2 8 7 9 2 100: ziczac
34 2 56 7 90 20 30 7 120: ziczac

```python
# Nhập danh sách từ bàn phím
ds = list(map(int, input().split()))

# Xử lý nhanh các trường hợp mảng quá ngắn (dưới 3 phần tử)
if len(ds) == 2 and ds[0] == ds[1]:
    print("khong ziczac")
elif len(ds) < 3:
    print("ziczac")
else:
    # Bắt đầu quét các phần tử ở giữa
    for i in range(1, len(ds) - 1):
        
        la_dinh_nui = (ds[i] > ds[i - 1]) and (ds[i] > ds[i + 1])
        la_thung_lung = (ds[i] < ds[i - 1]) and (ds[i] < ds[i + 1])
        
        # Nếu phát hiện một kẻ phá bĩnh (không phải đỉnh cũng không phải thung lũng)
        if not (la_dinh_nui or la_thung_lung):
            print("khong ziczac")
            break   # Kích hoạt break -> Lập tức dừng vòng lặp và BỎ QUA lệnh else bên dưới
            
    # Lệnh else này THUỘC VỀ VÒNG LẶP FOR (lùi vào cùng hàng với chữ for)
    else:
        # Chỉ khi vòng lặp for kiểm tra xong mọi thứ trót lọt (không bị break) 
        # thì máy tính mới in ra dòng này.
        print("ziczac")
```

### Bài 9: Tìm và in ra tất cả các vị trí của số nguyên k trong một danh sách số nguyên cho trước. Nếu không có k trong ds thì in ra "-1".
VD: k = 3, [1,3,4,4,3,3,4,3,6,1] => k xuất hiện tại các vị trí (trả về 1 list) [1, 4, 5, 7]
k = 10, [1,3,4,4,3,3,4,3,6,1] => k không xuất hiện trong ds => trả về "-1"

```python
def findk(lst, k):
    vtk = []
    for i in range(len(lst)):
        if lst[i] == k:
            vtk.append(i)  
            
    # Bước 2: Vòng lặp chạy xong hết rồi mới bắt đầu kiểm tra rổ
    if len(vtk) == 0:      # Nếu rổ trống không (tức là không có k)
        return "-1"
    else:                  # Nếu rổ có đồ
        return vtk

lst = list(map(int, input().split()))
k = int(input())
print(findk(lst, k))
    
```
